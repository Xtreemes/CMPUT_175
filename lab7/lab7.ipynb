### Q1: Load the Dataset
Load the dataset titanic.csv using appropriate data structure. Do not use any other library like csv, pandas, numpy, etc.

Print the total number of rows (including the header) in the dataset.



```python
def read_file(filename):
    """
    Opens a file and returns an array (two dimensional if 2+ columns).
    Splits lines by comma and trims any leading or trailing whitespaces.
    """

    with open(filename) as file:
        lines = file.read().splitlines()

        # check if further processing for 2d arrays is required
        if "," in lines[0]:
            for k, v in enumerate(lines):
                split_line = v.split(",")

                for i, entry in enumerate(split_line):
                    split_line[i] = entry.strip()

                lines[k] = split_line

        return lines

data = read_file("titanic.csv")
print(f"Total rows in dataset: {len(data)}")
```

### Q4: Filter the Dataset
Remove all unnecessary columns and retain only the relevant columns: Survived, Pclass, Sex, and Age. Print the new header and the first 10 rows of the filtered dataset.

```python
relevant_columns = ("Survived", "Pclass", "Sex", "Age")

column = 0
new_header = []
for header in headers:
    if header in relevant_columns:
        column += 1
        new_header.append(header)
    else:  # delete column
        for row in data:
            del row[column]

        if header == "Name":  # name gets split into 2 columns, delete twice
            for row in data:
                del row[column]

headers = new_header

print("Filtered Header:")
print(headers)

print("\nFirst 10 rows after filtering:")
for i in range(10):
    print(data[i])
```
